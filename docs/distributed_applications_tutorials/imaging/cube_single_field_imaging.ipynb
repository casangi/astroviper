# Single-Field Cube Imaging with AstroVIPER

**AstroVIPER** (Astro **V**isibility and **I**mage **P**arallel **E**xecution **R**eduction)
is the *science* layer of a three-package radio-astronomy stack:

| Package | Role |
| --- | --- |
| **AstroVIPER** | Science: gridding, weighting, FFT, deconvolution (CLEAN), imaging workflows |
| **[XRADIO](https://xradio.readthedocs.io/en/latest/)** | I/O + data model: Processing Sets, Measurement Set v4, image datasets |
| **[GraphVIPER](https://graphviper.readthedocs.io/en/latest/)** | Concurrency: builds and runs Dask MapReduce graphs |

This tutorial uses the **single-field cube imager** — `image_cube_single_field` — as a worked
example to show both *how to run* AstroVIPER and *how to write code for it*:

1. Run the imager end-to-end on ALMA TW Hydrae data (dirty image, then a multi-cycle CLEAN).
2. Tour the inputs and outputs: Processing Sets, image datasets, and **data groups**.
3. Understand the four-layer architecture the imager is built on.
4. Learn the coding patterns (with a small runnable example) for adding your own functionality.

**Prerequisites**: `pip install astroviper` (builds the C++ extensions, so it needs a compiler;
see the repository README for developer installs). The dataset used here is the same
5-channel TW Hydrae processing set used by the stakeholder tests in
`tests/stakeholder/test_single_field_imaging.py`.

## 2. Get the data

The input is a **Processing Set**: a Zarr store holding one or more Measurement Set v4s.
This one contains TW Hydrae ALMA data, split to 5 LSRK channels. It is distributed as a
zipped Zarr directory on Google Drive (the download is skipped when the directory is
already present).

````python
import glob
import os
import shutil
import zipfile

PS_STORE = "twhya_selfcal_lsrk_5chans.ps.zarr"
PS_STORE_DRIVE_ID = "1BRe3cD6YAWkn-jSPbClGGM9VlbxHP_yn"


def download_zarr_from_drive(zarr_name, file_id):
    """Download and extract one zipped ``.zarr`` directory from Google Drive.

    A no-op when the directory already exists locally. Handles archives that
    arrive double-zipped (a ``.zip`` whose only member is another ``.zip``).
    """
    if os.path.isdir(zarr_name):
        return  # already present locally -- nothing to download.

    import gdown

    zip_path = zarr_name + ".zip"
    gdown.download(id=file_id, output=zip_path, quiet=False)

    work_dir = zarr_name + ".extract"
    shutil.rmtree(work_dir, ignore_errors=True)
    os.makedirs(work_dir)
    shutil.move(zip_path, os.path.join(work_dir, os.path.basename(zip_path)))

    for _ in range(6):  # safety bound against malformed archives
        for root, dirs, _files in os.walk(work_dir):
            if zarr_name in dirs:
                shutil.move(os.path.join(root, zarr_name), zarr_name)
                shutil.rmtree(work_dir, ignore_errors=True)
                return
        nested_zips = glob.glob(os.path.join(work_dir, "**", "*.zip"), recursive=True)
        if not nested_zips:
            break
        for nested in nested_zips:
            with zipfile.ZipFile(nested) as zf:
                zf.extractall(os.path.dirname(nested))
            os.remove(nested)

    shutil.rmtree(work_dir, ignore_errors=True)
    raise RuntimeError(f"Could not extract '{zarr_name}' from its archive.")


download_zarr_from_drive(PS_STORE, PS_STORE_DRIVE_ID)
````

## 3. Explore the Processing Set

A Processing Set (`ps_xdt`) is an `xarray.DataTree` whose children are individual
**Measurement Set v4s** — self-describing correlated data for one observation / spectral
window / polarization setup, with dimensions `time x baseline_id x frequency x polarization`.

Two naming conventions from XRADIO to remember:

- `open_*` functions are **lazy** (metadata only; data variables are lazily loaded), while
  `load_*` functions are **eager** (everything into memory now). The driver opens the
  processing set lazily and each parallel task loads only its own chunk.
- **Coordinates** are lowercase (`frequency`, `time`, `polarization`); **data variables** are
  UPPERCASE (`VISIBILITY`, `WEIGHT`, `FLAG`, `UVW`).

The `xr_ps` accessor provides processing-set-level helpers such as `summary()`.

````python
from xradio.measurement_set import open_processing_set

ps_xdt = open_processing_set(PS_STORE)
ps_xdt.xr_ps.summary()
````

### Data groups

A **data group** lets one dataset hold multiple versions of the same logical variable
(raw vs. corrected vs. model visibilities, say) without overwriting anything. It is a dict
stored in `attrs["data_groups"]` mapping a group name to *logical roles* (lowercase keys:
`correlated_data`, `flag`, `weight`, `uvw`) whose values are *data-variable names*
(UPPERCASE). Groups may share variables. Below, the imager is pointed at a group with
`processing_set_data_group_name`; everything it reads (visibilities, flags, weights, uvw)
is resolved through that group.

````python
ms_name, ms_xdt = list(ps_xdt.items())[0]
print(f"Measurement set: {ms_name}")
ms_xdt.attrs["data_groups"]
````

## 4. Imaging parameters

The driver takes three parameter dicts (validated against a JSON schema that lives next to
the driver module — more on that in section 8):

**`image_params`** — the image geometry and output coordinates:

- `image_size` — pixels along `l` and `m`.
- `cell_size` — pixel size in **radians**; the first (l / right-ascension) element is negative
  by convention so RA increases leftward.
- `phase_direction` — image phase center; taken from the processing set's combined
  field-and-source dataset below.
- `frequency_coords` — output cube channels; here simply the processing set's frequency axis.
- `polarization_coords` — the **Stokes** basis of the output image (gridding itself happens in
  the instrument's correlation basis — `XX`/`YY` for ALMA's linear feeds — controlled by the
  separate `instrument_polarization_basis` argument).
- `time_coords` — output time axis (`[0]` collapses all times into one plane).
- `fft_padding` — padding factor applied to the uv grid before the FFT.
- `cpp_gridder` — choose the C++ gridding kernel (`True`) or its Numba twin (`False`); the two
  are kept mathematically interchangeable.

**`imaging_weights_params`** — `"natural"` or `"briggs"` weighting (with the usual `robust`
parameter).

**`iteration_control_params`** — the CLEAN controls, matching CASA `tclean` semantics with one
deliberate difference: iteration control is **independent per (time, frequency, polarization)
plane**. Each plane has its own `niter` budget and `threshold`, and major cycles continue until
*every* plane has stopped. The main knobs: `niter` (max CLEAN components per plane), `nmajor`
(max major cycles; `-1` = unlimited), `threshold` (Jy, absolute stop), `gain` (loop gain),
`primary_beam_limit` (mask cutoff as a fraction of the peak primary beam), and
`cyclefactor` / `cycleniter` / `minpsffraction` / `maxpsffraction` (when a minor cycle hands
control back to the next major cycle).

````python
import numpy as np

combined_field_xds = ps_xdt.xr_ps.get_combined_field_and_source_xds()
center_field_name = combined_field_xds.attrs["center_field_name"]
phase_direction = combined_field_xds.FIELD_PHASE_CENTER_DIRECTION.sel(
    field_name=center_field_name
)

image_params = {
    "image_size": [250, 250],
    "cell_size": np.array([-0.1, 0.1]) * np.pi / (180 * 3600),  # 0.1 arcsec pixels
    "phase_direction": phase_direction.values,
    "frequency_coords": ps_xdt.xr_ps.get_freq_axis().values,
    "polarization_coords": ["I", "Q"],
    "time_coords": [0],
    "fft_padding": 1.2,
    "cpp_gridder": True,
}

imaging_weights_params = {
    "weighting": "briggs",
    "robust": 0.5,
    "casa_weighting_implementation": True,
}
````

## 5. Make a dirty image (`niter=0`)

With `niter=0` the imager performs no deconvolution: it computes imaging weights, grids the
visibilities, and FFTs them into the **dirty image** (stored as `SKY_RESIDUAL` — with no model
subtracted, the residual *is* the dirty image), along with the **point spread function** and
the **primary beam**.

`image_data_variables_keep` selects which image variables are written to the output Zarr
store; everything else is discarded when the per-chunk task finishes. `n_chunks` sets how many
frequency chunks the cube is split into for parallel processing (one Dask task each;
`None` auto-picks from a per-channel memory estimate), while `processing_function_threads` is
the *within*-task thread count handed to the C++/Numba/FFT kernels — the two levels of
parallelism multiply, so keep `n_chunks x processing_function_threads` near your core count.

````python
from astroviper.distributed_applications.imaging import image_cube_single_field

DIRTY_IMAGE_STORE = "twhya_dirty.img.zarr"

dirty_iteration_control = {
    "niter": 0,
    "nmajor": 0,
    "threshold": 0.0,
    "gain": 0.1,
    "cyclefactor": 1.5,
    "cycleniter": -1,
    "minpsffraction": 0.05,
    "maxpsffraction": 0.8,
}

return_dict_dirty = image_cube_single_field(
    ps_store=PS_STORE,
    image_store=DIRTY_IMAGE_STORE,
    image_params=image_params,
    imaging_weights_params=imaging_weights_params,
    iteration_control_params=dirty_iteration_control,
    gridder="prolate_spheroidal",
    deconvolver="hogbom_many_threads",
    scan_intents="OBSERVE_TARGET#ON_SOURCE",
    image_data_variables_keep=[
        "sky_residual",
        "point_spread_function",
        "primary_beam",
        "beam_fit_params_point_spread_function",
    ],
    processing_set_data_group_name="base",
    single_precision_image=False,
    processing_function_threads=1,
    n_chunks=5,
    overwrite=True,
)
````

The returned dict contains three things: the merged per-plane deconvolution statistics
(`"deconvolution"`, empty here since `niter=0`), a `pandas.DataFrame` of per-chunk node-task
timings (`"timing_node_tasks"`), and the driver-level step timings
(`"timing_distributed_application"`).

````python
return_dict_dirty["timing_node_tasks"]
````

## 6. CLEAN: major and minor cycles

Now a real deconvolution. Each **major cycle** degrids the current model into model
visibilities, subtracts them from the observed visibilities *in double precision*, and grids
and FFTs the difference into a residual image. Each **minor cycle** (the deconvolver — here
Hogbom CLEAN) then picks peaks in the residual, inside a mask built from the primary-beam
limit, and moves flux into the model image until `cycleniter` / `cyclethreshold` triggers the
next major cycle. A final residual cycle always follows the last minor cycle so the reported
residual is consistent with the final model.

With `restore=True` the model is convolved with the **clean beam** (the per-channel Gaussian
fit to the PSF, stored in `BEAM_FIT_PARAMS_POINT_SPREAD_FUNCTION`) and added to the residual,
producing `SKY_RESTORED`.

Two things worth noting in the call below:

- `single_precision_image=True` (the default) keeps the *image-domain* arrays — uv grids and
  sky/PSF/model images — in single precision, halving the cube's memory footprint. The
  visibilities always stay `complex128`, and the residual is always formed in double
  precision.
- `sky_model` and `mask` are added to the keep list so the CLEAN products are written out.

````python
CLEAN_IMAGE_STORE = "twhya_clean.img.zarr"

clean_iteration_control = {
    "niter": 300,
    "nmajor": 3,
    "threshold": 0.001,  # Jy
    "primary_beam_limit": 0.2,
    "gain": 0.1,
    "cyclefactor": 1.5,
    "cycleniter": -1,
    "minpsffraction": 0.05,
    "maxpsffraction": 0.8,
}

return_dict_clean = image_cube_single_field(
    ps_store=PS_STORE,
    image_store=CLEAN_IMAGE_STORE,
    image_params=image_params,
    imaging_weights_params=imaging_weights_params,
    iteration_control_params=clean_iteration_control,
    gridder="prolate_spheroidal",
    deconvolver="hogbom_many_threads",
    scan_intents="OBSERVE_TARGET#ON_SOURCE",
    image_data_variables_keep=[
        "sky_residual",
        "sky_model",
        "mask",
        "point_spread_function",
        "primary_beam",
        "beam_fit_params_point_spread_function",
    ],
    processing_set_data_group_name="base",
    single_precision_image=True,
    processing_function_threads=1,
    n_chunks=5,
    overwrite=True,
    restore=True,
)
````

`return_dict["deconvolution"]` is a `ReturnDict` of per-plane convergence statistics keyed by
`(time, polarization, channel)`: iterations done, peak residual per cycle, the mask size, and
a CASA-style stop code explaining *why* each plane stopped.

````python
from astroviper.processing_functions.imaging.utils import print_deconvolve_dict

print_deconvolve_dict(return_dict_clean["deconvolution"])
````

````python
clean_xds = xr.open_zarr(CLEAN_IMAGE_STORE)
print("Image data groups:")
for name, group in clean_xds.attrs["data_groups"].items():
    print(f"  {name}: {group}")

chan = 2
fig, axes = plt.subplots(1, 3, figsize=(15, 4.2), constrained_layout=True)
for ax, var in zip(axes, ["SKY_RESIDUAL", "SKY_MODEL", "SKY_RESTORED"]):
    plane = clean_xds[var].isel(time=0, frequency=chan, polarization=0)
    im = ax.imshow(plane.values, origin="lower", cmap="viridis")
    ax.set_title(f"{var}\nchannel {chan}, Stokes I")
    fig.colorbar(im, ax=ax)
plt.show()
````

## 7. How it works: the four-layer architecture

Everything you just ran is organized into four layers. **Calls only ever go downward**, and
each layer has a strict responsibility (this separation is a hard design rule of the stack —
concurrency lives in GraphVIPER, science in AstroVIPER processing functions, and they never
mix):

```
src/astroviper/
|-- distributed_applications/  (1) Build + compute Dask graphs (GraphVIPER map/reduce).
|                                  Nodes are node_tasks. NO science here.
|-- node_tasks/                (2) Thin per-chunk functions. Do I/O (load the chunk,
|                                  write Zarr) + call processing_functions.
|-- processing_functions/      (3) The science. Pure Python / Numba / C++ (pybind11).
|                                  In-memory xarray/NumPy only. NO graph/Dask code, no
|                                  processing-set I/O.
|-- utils/                     (4) Helpers: data_group_tools, io, data_partitioning, ...
```

The single-field imager exists at the top three layers, all named
`image_cube_single_field`:

1. **Driver** — `distributed_applications/imaging/image_cube_single_field.py` (what you
   called above). It creates the empty image store, decides the frequency chunking, builds
   `parallel_coords`, pre-allocates the output Zarr variables so each task can write its own
   slice, then builds and computes the graph:
   `map(...)` -> `reduce(...)` -> `generate_dask_workflow(...)` -> `dask.compute(...)`.
2. **Node task** — `node_tasks/imaging/image_cube_single_field.py`. Runs once per frequency
   chunk: loads that chunk of the processing set (eagerly), builds the per-chunk empty image
   in the instrument's correlation basis, calls the science layer, and writes the finished
   slice into the image store.
3. **Processing function** — `processing_functions/imaging/image_cube_single_field.py`. The
   actual science: the major/minor-cycle loop over gridding, FFT + normalization,
   deconvolution, and polarization transforms, all on in-memory arrays.

Two GraphVIPER details make this pleasant to work with:

- **Explicit node-task signatures.** GraphVIPER invokes node tasks with a single
  `input_params` dict, but `graphviper.graph_tools.map` automatically adapts any node task
  with a normal, fully documented keyword signature — so all three layers remain real,
  directly-callable functions.
- **I/O happens inside the compute nodes** (not as separate Dask array tasks). This keeps the
  graph small and sidesteps Dask memory backpressure on large image cubes — a deliberate
  design decision; keep it that way when adding workflows.

There are **two levels of parallelism**: across chunks (the Dask graph; `n_chunks`) and within
a task (`processing_function_threads`, passed as `num_threads` to the C++/Numba kernels).

One more contract worth knowing before writing kernels — the **Python <-> C++ memory
contract**: Python owns all large-array memory; C++ kernels receive pointers to Python-owned
NumPy buffers, verify dtype/shape/contiguity (no silent copies — never `forcecast` a large
array), write results in place, and release the GIL around the hot loop. The Numba kernels
follow the same in-place, `nogil=True` philosophy.

## 8. Writing your own AstroVIPER code

### The processing-function pattern

A processing function is a stateless science function on in-memory data. The standard shape,
used everywhere in `processing_functions/imaging/`, is a *data-group two-step*: resolve the
input/output groups up front (this validates the input exists and guards against accidental
overwrites), compute and write the new data variable, then register the output group with an
audit trail. Never mutate `attrs["data_groups"]` by hand.

Here is a small but complete example following the pattern — it smooths the model image with
a Gaussian kernel and registers the result as a new `model_smoothed` data group. The cube on
disk does not register its products in `data_groups`, so the example first registers a
`model` group itself with `modify_data_groups_xds` — which also shows how a group is created
in the first place:

````python
from astroviper.utils.data_group_tools import (
    create_data_groups_in_and_out,
    modify_data_groups_xds,
)


def smooth_sky_single_field(
    img_xds,
    sigma_pixels=2.0,
    image_data_group_in_name="model",
    image_data_group_out_name="model_smoothed",
    image_data_group_out_modified={"sky": "SKY_MODEL_SMOOTHED"},
    overwrite=False,
):
    """Gaussian-smooth the sky image of one data group into a new data group.

    Parameters
    ----------
    img_xds : xarray.Dataset
        Image dataset with dims ``(time, frequency, polarization, l, m)``.
    sigma_pixels : float
        Standard deviation of the Gaussian kernel, in pixels.
    image_data_group_in_name : str
        Data group whose ``sky`` variable is smoothed.
    image_data_group_out_name : str
        Name of the data group registered for the smoothed image.
    image_data_group_out_modified : dict
        Role -> new data-variable name for the output group.
    overwrite : bool
        Allow overwriting an existing output group.

    Returns
    -------
    xarray.Dataset
        ``img_xds`` with the smoothed variable added (modified in place).
    """
    from scipy.ndimage import gaussian_filter

    # 1) Resolve the in/out data groups (validates input, guards overwrite).
    data_group_in, data_group_out = create_data_groups_in_and_out(
        img_xds,
        data_group_in_name=image_data_group_in_name,
        data_group_out_name=image_data_group_out_name,
        data_group_out_modified=image_data_group_out_modified,
        overwrite=overwrite,
    )

    # 2) Compute and write the new data variable (smooth only over l, m).
    sky = img_xds[data_group_in["sky"]]
    smoothed = gaussian_filter(
        sky.values, sigma=(0, 0, 0, sigma_pixels, sigma_pixels)
    )
    img_xds[data_group_out["sky"]] = (sky.dims, smoothed.astype(sky.dtype))

    # 3) Register the new group (in place) with an audit-trail description.
    modify_data_groups_xds(
        img_xds,
        data_group_out_name=image_data_group_out_name,
        data_group_out=data_group_out,
        description=f"Model image smoothed with a {sigma_pixels}-pixel Gaussian.",
    )
    return img_xds


img_xds = xr.open_zarr(CLEAN_IMAGE_STORE).load()

# Register a data group for the model image the imager wrote, so the
# data-group tooling can resolve it as an input.
modify_data_groups_xds(
    img_xds,
    data_group_out_name="model",
    data_group_out={"sky": "SKY_MODEL"},
    description="CLEAN model image produced by image_cube_single_field.",
)

img_xds = smooth_sky_single_field(img_xds, sigma_pixels=2.0)

for name, group in img_xds.attrs["data_groups"].items():
    print(f"{name}: {group}")
````

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4.2), constrained_layout=True)
for ax, var in zip(axes, ["SKY_MODEL", "SKY_MODEL_SMOOTHED"]):
    plane = img_xds[var].isel(time=0, frequency=2, polarization=0)
    im = ax.imshow(plane.values, origin="lower", cmap="viridis")
    ax.set_title(var)
    fig.colorbar(im, ax=ax)
plt.show()
````

Conventions the example follows (and yours should too):

- **NumPy-style docstring** on every public function.
- Data-group parameters use the standard vocabulary: `image_data_group_in_name` /
  `image_data_group_out_name` / `image_data_group_out_modified` on the image side, and the
  `ms_data_group_*` equivalents on the measurement-set side.
- New variable versions are the standard name plus a descriptor:
  `SKY_MODEL` -> `SKY_MODEL_SMOOTHED`.
- Arrays are modified **in place**; the function returns stats/timings (or the dataset), not
  copies of large arrays.

### The node-task pattern

To run a processing function chunk-parallel, wrap it in a **node task** in `node_tasks/` — a
thin function with an explicit keyword signature that (1) loads its chunk of the input (or
uses the pre-loaded `input_data` the data-loading layer injects), (2) calls the processing
function, and (3) writes the result slice to the output Zarr store
(`astroviper.utils.io.write_result_chunk_to_disk_using_zarr`). Keep node tasks thin: I/O and
orchestration only.

### The driver pattern

Finally, a **driver** in `distributed_applications/` makes it a distributed application
(sketch of the essential steps, following the imager driver):

```python
from graphviper.graph_tools.coordinate_utils import (
    make_parallel_coord,
    interpolate_data_coords_onto_parallel_coords,
)
from graphviper.graph_tools import map, reduce, generate_dask_workflow
import dask

parallel_coords = {"frequency": make_parallel_coord(coord=img_xds.frequency, n_chunks=n_chunks)}
node_task_data_mapping = interpolate_data_coords_onto_parallel_coords(parallel_coords, ps_xdt)

viper_graph = map(
    input_data=ps_xdt,
    node_task_data_mapping=node_task_data_mapping,
    node_task=my_node_task,          # explicit signature; map() adapts it automatically
    input_params=input_params,       # everything the node task needs, plus framework keys
)
viper_graph = reduce(viper_graph, my_reduce_fn, input_params, mode="tree")
dask_graph = generate_dask_workflow(viper_graph)
results = dask.compute(dask_graph)
```

Driver-level parameters are validated **once, at this layer**, with toolviper's schema
validation: decorate the driver with
`@toolviper.utils.parameter.validate(config_dir=os.path.dirname(os.path.abspath(__file__)))`
and put the `<module_name>.param.json` schema **next to the module**. Node tasks and
processing functions trust the already-validated parameters.

### Where a change goes

| Task | Where |
| --- | --- |
| New workflow / parallelism | `distributed_applications/<subdomain>/` |
| New per-chunk I/O + orchestration | `node_tasks/<subdomain>/` |
| New science (gridding, weighting, deconvolution, ...) | `processing_functions/<subdomain>/` |
| Data-group helpers | `utils/data_group_tools.py` |
| Zarr variable definitions / chunk writers | `utils/io.py` |
| Unit tests (mirror the `src/` layout) | `tests/unit/...` |

## 9. Checklist and further reading

Before opening a pull request:

- **Naming**: lowercase `snake_case` coordinates, UPPERCASE data variables, descriptive
  function names (`image_size`, not `imsize`).
- **Formatting**: run **black** (CI enforces it); notebook outputs are stripped by the
  `nbstripout` pre-commit hook.
- **Layering**: no Dask in `processing_functions/`, no science in
  `distributed_applications/`.
- **Memory contract**: no `forcecast` / silent copies of large arrays across the Python <-> C++
  boundary; kernels write in place and release the GIL.
- **Tests**: add or update unit tests under the mirrored path in `tests/unit/`; end-to-end
  science validation lives in `tests/stakeholder/`.
