This notebook demonstrates how to create and update a mask image using the `make_mask` function. The function takes an xradio image `xarray.Dataset` (`img_xds`) that contains a primary-beam data variable referenced by an `"image"` data group in `img_xds.attrs["data_groups"]`, and writes a boolean mask data variable back into the same dataset (in place). The current implementation supports a primary-beam threshold mask for deconvolution; other mask types may be added in the future.

---

---
## Notes

- The input `img_xds` is expected to follow the xradio image schema and to expose a `data_groups` attribute mapping role names (`primary_beam`, `mask`, ...) to data-variable names.
- `make_mask` modifies `img_xds` in place and returns `None`. The new mask data variable is added (or overwritten when `overwrite=True`) and the output data group is registered on `img_xds.attrs["data_groups"]` with a UTC timestamp and a description.
- To use `combine_mask=True`, the input data group must include a `"mask"` entry pointing to an existing mask variable, and `overwrite=True` is required when the output mask variable name collides with the input mask variable name.
- Only a primary-beam threshold mask is currently supported; auto-masking and other mask types will be added later.

## Import additional modules/functions used in this notebook

```python
from datetime import datetime, timezone

import dask.array as da
import matplotlib.pyplot as plt
import numpy as np
import xarray as xr
from astropy.modeling.models import Gaussian2D
from astropy.time import Time
from xradio.image import make_empty_sky_image
```

## Generate input images

The helper below builds an ALMA mock-up image `xarray.Dataset` for one of two roles:

- `image_type='pb'` — Gaussian primary beam stored as a `PRIMARY_BEAM` data variable with `data_groups = {"image": {"primary_beam": "PRIMARY_BEAM"}}`.
- `image_type='residual'` — random residual stored as `SKY` with `data_groups = {"image": {"sky": "SKY"}}`. When `add_mask=True`, an arbitrary rectangular mask is added as `MASK` and registered under the `"mask"` role.

`make_mask` ultimately needs the primary beam and (for `combine_mask=True`) any existing mask to live on the *same* `img_xds`; the cells further down combine these helpers as needed.

```python
def generate_image_xds(
    phase_center, image_size, cell_size, freqs, image_type, add_mask=False
):
    # Set polarization to I and a single time stamp.
    pols = np.array(["I"])
    t = Time(datetime(2025, 10, 1, 0, 0))
    times = np.array([t.mjd])

    # Build an empty sky image with the requested geometry/coords.
    new_xds = make_empty_sky_image(
        phase_center=phase_center,
        image_size=image_size,
        cell_size=cell_size,
        frequency_coords=freqs,
        pol_coords=pols,
        time_coords=times,
        do_sky_coords=True,
    )
    shape = tuple(
        new_xds.sizes[d] for d in ["time", "frequency", "polarization", "l", "m"]
    )
    y = new_xds.l.data
    x = new_xds.m.data
    xv, yv = np.meshgrid(x, y)

    image_data = np.zeros(shape)

    if image_type == "pb":
        # Synthesise an ALMA 12 m Gaussian primary beam per channel.
        dish_dia = 12.0
        for index, freq in enumerate(freqs):
            sigma = (1.13 / 2.355) * (2.99792e8 / (freq * dish_dia))
            pbmodel = Gaussian2D(x_stddev=sigma, y_stddev=sigma)
            pb_data = pbmodel(xv, yv)
            image_data[0, index, 0] = da.from_array(pb_data, chunks=(25, 25))

        # New API: store under PRIMARY_BEAM and reference it in the
        # "image" data group under the "primary_beam" role.
        new_xds["PRIMARY_BEAM"] = (
            ("time", "frequency", "polarization", "l", "m"),
            image_data,
        )
        new_xds.attrs["data_groups"] = {"image": {"primary_beam": "PRIMARY_BEAM"}}

    elif image_type == "residual":
        # Stand-in residual: random noise per channel.
        for index, freq in enumerate(freqs):
            image_data[0, index, 0] = da.random.normal(
                -1, 1, tuple(image_size), chunks=(25, 25)
            )

        new_xds["SKY"] = (
            ("time", "frequency", "polarization", "l", "m"),
            image_data,
        )
        data_group = {"sky": "SKY"}

        if add_mask:
            # Arbitrary rectangular boolean mask to demonstrate combine_mask=True.
            mask_data = np.zeros(shape, dtype=bool)
            mask_data[
                :,
                :,
                :,
                int(image_size[0] * 0.2) : int(image_size[0] * 0.8),
                int(image_size[1] * 0.5) : int(image_size[1] * 0.9),
            ] = True
            new_xds["MASK"] = (
                ("time", "frequency", "polarization", "l", "m"),
                da.from_array(mask_data, chunks="auto"),
            )
            data_group["mask"] = "MASK"

        new_xds.attrs["data_groups"] = {"image": data_group}

    return new_xds
```

```python
# Image parameters
# phase center and cell_size are in radians
phase_center = [np.float64(1.4092068104521434), np.float64(-0.6363220835781705)]
image_size = np.array([100, 100])
cell_size = np.array([4.84813681109536e-06, 4.84813681109536e-06])
# channel frequencies in Hz
freqs = np.array([1.0e11, 1.2e11])
```

### Generate PB xds @~100GHz

```python
pb_xds = generate_image_xds(
    phase_center=phase_center,
    image_size=image_size,
    cell_size=cell_size,
    freqs=freqs,
    image_type="pb",
)
```

### Generate residual image with an internal mask

```python
residual_xds = generate_image_xds(
    phase_center=phase_center,
    image_size=image_size,
    cell_size=cell_size,
    freqs=freqs,
    image_type="residual",
    add_mask=True,
)
```

```python
nchan = len(freqs)
fig, axes = plt.subplots(2, 3, figsize=(10, 8))
for i in range(nchan):
    data_sel = {"time": 0, "frequency": i, "polarization": 0}
    pb_xds["PRIMARY_BEAM"].isel(data_sel).squeeze().plot.pcolormesh(
        x="right_ascension",
        y="declination",
        ax=axes[i, 0],
        cmap="grey",
        add_colorbar=False,
    )
    axes[i, 0].set_title(f"PB chan {i}")
    axes[i, 0].set_aspect(1.3)

    residual_xds["SKY"].isel(data_sel).squeeze().plot.pcolormesh(
        x="right_ascension",
        y="declination",
        ax=axes[i, 1],
        cmap="viridis",
        add_colorbar=False,
    )
    axes[i, 1].set_title(f"residual chan {i}")
    axes[i, 1].yaxis.set_visible(False)
    axes[i, 1].set_aspect(1.3)

    residual_xds["MASK"].isel(data_sel).squeeze().plot.pcolormesh(
        x="right_ascension",
        y="declination",
        ax=axes[i, 2],
        cmap="grey",
        add_colorbar=False,
    )
    axes[i, 2].set_title(f"in-mask chan {i}")
    axes[i, 2].yaxis.set_visible(False)
    axes[i, 2].set_aspect(1.3)

axes[0, 0].xaxis.set_visible(False)
axes[0, 1].xaxis.set_visible(False)
axes[0, 2].xaxis.set_visible(False)
plt.show()
```

## Example 1: Create a PB threshold-based mask

`make_mask` modifies `img_xds` in place: it adds a boolean `MASK` data variable and updates `img_xds.attrs["data_groups"]` with the resolved output group (including a UTC timestamp and a description). Here we work on a copy of `pb_xds` so subsequent examples can reuse the original.

### Create the mask

```python
img_xds = pb_xds.copy()
make_mask(img_xds, primary_beam_limit=0.9)
```

## Example 2: Combining with an existing mask

To combine a new PB-threshold mask with an existing mask, both the primary beam and the existing mask must live on the same `img_xds` and be referenced by the input data group. Below we build such a combined dataset by attaching `PRIMARY_BEAM` from `pb_xds` to a copy of `residual_xds` (which already carries `MASK` plus the `"image"` data group with the `mask` role).

`combine_mask=True` writes the logical OR of the existing mask and the threshold mask back into the output mask variable. Because the output mask name (`"MASK"`) collides with the input mask name, `overwrite=True` is required.

```python
combined_xds = residual_xds.copy()
combined_xds["PRIMARY_BEAM"] = pb_xds["PRIMARY_BEAM"]

# Extend the input data group with the primary_beam role so make_mask can
# find the PB variable.  The existing "mask" entry remains in place.
combined_xds.attrs["data_groups"] = {
    "image": {
        **combined_xds.attrs["data_groups"]["image"],
        "primary_beam": "PRIMARY_BEAM",
    }
}

make_mask(combined_xds, primary_beam_limit=0.5, combine_mask=True, overwrite=True)
```

### Plot the combined mask

```python
fig, axes = plt.subplots(1, 2)
for i in range(len(freqs)):
    data_sel = {"time": 0, "frequency": i, "polarization": 0}
    combined_xds["MASK"].isel(data_sel).squeeze().plot.pcolormesh(
        x="right_ascension",
        y="declination",
        ax=axes[i],
        cmap="grey",
        add_colorbar=False,
    )
    axes[i].set_title(f"combined mask chan {i}")
    axes[i].set_aspect(1.3)
axes[-1].yaxis.set_visible(False)
plt.show()
```

## Example 3: Storing the result under a separate output data group

By default `make_mask` registers the output under the same data-group key it read from (`image_data_group_out_name="image"`), updating that group's audit trail. Passing a different `image_data_group_out_name` keeps the input group untouched and adds a new group entry — useful when you want to retain the prior pipeline stage's metadata alongside the new one.

```python
img_xds_v2 = pb_xds.copy()
make_mask(
    img_xds_v2,
    primary_beam_limit=0.9,
    image_data_group_out_name="image_masked",
    image_data_group_out_modified={"mask": "MASK_PB"},
)
```

### Both data groups are now registered, and the new mask is in `MASK_PB`.
